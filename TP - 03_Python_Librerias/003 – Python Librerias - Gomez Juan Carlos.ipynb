Ejercicio 2 – Dataset simple con Pandas
Crear un DataFrame de Pandas con la siguiente información:
estudiante	nota
Ana	                  8
Juan	          6
Carla	          9
Pedro	          5
Lucia	          7
Luego realizar las siguientes operaciones:
Mostrar el DataFrame
Calcular el promedio de las notas
Mostrar la nota máxima
Mostrar la nota mínima

```
# Ejercicio 2 - Dataset simple con Pandas
import pandas as pd

datos = {"estudiante": ["Ana", "Juan", "Carla", "Pedro", "Lucia"], 
        "nota": [8, 6, 9, 5, 7]
        }

df = pd.DataFrame(datos)

# Mostrar el DataFrame
print(df)

# Calcular el promedio
promedio = df["nota"].mean()
print("Promedio de las notas:", promedio)

# Mostrar la nota maxima
nota_maxima = df["nota"].max()
print("Nota maxima:", nota_maxima)

# Mostrar la nota minima
nota_minima = df["nota"].min()
print("Nota minima:", nota_minima)
```
