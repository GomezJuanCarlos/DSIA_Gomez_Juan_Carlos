Ejercicio 1 – Variables y operaciones
Crear variables para almacenar:
•	tu nombre
•	tu edad
•	tu ciudad
Luego imprimir un mensaje con el siguiente formato:
Hola, mi nombre es \_\_\_, tengo \_\_\_ años y vivo en \_\_\_\.
 A continuación, crear dos variables numéricas a y b, asignarles valores y calcular:
•	suma
•	resta
•	multiplicación
•	división
Mostrar los resultados utilizando print\(\)\.

```
# Ejercicio 1: Variables y operaciones
print(f"RESENTACION")

nombre = "Juan Carlos"
edad = 40
ciudad = "Gregorio de Laferrere"

print(f"Hola, mi nombre es {nombre}, tengo {edad} años y vivo en {ciudad}.\n")

print(f"OPERACIONES")

# Variables numericas
a = 21
b = 3

# Operaciones
suma = a + b
resta = a - b
multiplicacion = a * b
division = a / b

# Mostrar resultados
print("Suma:", suma)
print("Resta:", resta)
print("Multiplicación:", multiplicacion)
print("División:", division)
```

Ejercicio 2 – Condicionales
Crear un programa que:
1\.	Pida al usuario que ingrese su edad usando input\(\)
2\.	Determine si la persona es mayor o menor de edad
Mostrar uno de los siguientes mensajes:
Eres mayor de edad o Eres menor de edad

```
# Ejercicio 2
# Pedir la edad al usuario
edad = int(input("Ingresa tu edad: "))

# Determinar si es mayor o menor de edad
if edad >= 18:
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
```

Ejercicio 3 – Listas y bucles
Crear una lista con 5 números\. Utilizando un bucle for, imprimir cada número de la lista\.
Luego:
•	Calcular la suma total de los números
•	Calcular el promedio
Mostrar ambos resultados por pantalla\.

```
# Ejercicio 3
# Lista con 5 numeros
print(f"LISTA DE 5 NUMEROS")
numeros = [14, 21, 36, 44, 58]

# Imprimir cada nro de la lista
for numero in numeros:
    print(numero)

# Calcular la suma
suma = sum(numeros)

# Calcular el promedio
promedio = suma / len(numeros)

# Mostrar resultados
print("Suma total:", suma)
print("Promedio:", promedio)
```
